```python

import pandas as pd

df = pd.read_csv("E:\public-policy-data-project\data\processed\education-ptr.csv")

print("Rows and columns:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst five rows:")
display(df.head())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isna().sum())
```

```python

import pandas as pd

# Keep only the fields needed for this project.
columns = [
    "India/State/UT",
    "Pupil Teacher Ratio (PTR) - Primary (1 to 5)",
    "Pupil Teacher Ratio (PTR) - Upper Primary (6-8)",
    "Pupil Teacher Ratio (PTR) - Secondary (9-10)",
    "Pupil Teacher Ratio (PTR) - Higher Secondary (11-12)"
]

df = df[columns].copy()

# Standardize text and numeric fields.
df["India/State/UT"] = df["India/State/UT"].astype("string").str.strip()

for col in columns[1:]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Inspect data quality.
print("Duplicate state names:", df["India/State/UT"].duplicated().sum())
print(df.isna().sum())

# Summary statistics.
print(df[columns[1:]].describe())

# Compare states for primary education.
primary = df.dropna(subset=["Pupil Teacher Ratio (PTR) - Primary (1 to 5)"])

print(
    primary.sort_values("Pupil Teacher Ratio (PTR) - Primary (1 to 5)")
    [["India/State/UT", "Pupil Teacher Ratio (PTR) - Primary (1 to 5)"]]
    .to_string(index=False)
)
```
